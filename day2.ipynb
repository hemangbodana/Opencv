# `Image Rotation`
Process of rotating an image by certain angle. Useful when we align an image or when we want to view an image from different angles.

Can be done using `cv2.getRotationMatrix2D()` and `cv2.warpAffine()` functions.

`cv2.getRotationMatrix2D()` :- takes **center** of the rotation(specified as (cols/2, rows/2)) and **angle** of rotation in **degrees** and **scaling factor**(1=normal no zoom,>1 zoom in, <1 zoom out). It returns the affine matrix(M).

`cv2.warpAffine()` :- Used to perform affine transformation on an image. An Affine transformation is a linear mapping that preserves parallel lines and ratios of distances between points. It is used to rotate, scale, shear, and translate an image.
- It just applies the rotation matrix M to your image.
- It takes the **destination** image size as an argument.

```python
import cv2 
img = cv2.imread('apple.png')

# NOTE warpAffine expects sizes as (Width,Height) and not (Height,Width) but shape[::2] will return (Height,Width)

# Rotating the image
(rows,cols) = img.shape[:2]  #will return (640,480) instead of (640,480,3)
M = cv2.getRotationMatrix2D((cols/2,rows/2),40,2)
# print(M) Affine matrix

rotated_img= cv2.warpAffine(img,M,(cols,rows))
cv2.imshow('Rotated Image',rotated_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# `Image Split`
Process of splitting into its individual color channels and can be useful for a variety of image processing tasks, such as color separation, image enhancement, and image restoration.Can be done using **cv2.split()** function.

`cv2.split()` takes an **image** and returns a tuple containing the individual color channels as separate Grayscale images.

```python
import cv2
import numpy as np

# Loading the image
img = cv2.imread('apple.png')
# Spliting the image into its color channels
b, g, r = cv2.split(img)

# Displaying the image on individual channels
cv2.imshow('Blue Channel', b)
cv2.imshow('Green Channel', g)
cv2.imshow('Red Channel', r)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

# `Merging Image`
Process of merging multiple color channels or grayscale into a single multi-channel image. Can be done using **`cv2.merge`**.

**`cv2.merge`**- takes a tuple of individual grayscale or color channels as separate images and returns a single multi-channel image.

```python
import cv2 

img0 = cv2.imread("parrot.jpg")
cv2.imshow("Merged Image",img0)
cv2.waitKey(0)
cv2.destroyAllWindows()

# Mergin the image with another parrot
b,g,r = cv2.split(img0)
img0 = cv2.merge((b,g,r))
parrotface = img0[155:355,0:400] #ROW pe chalta he, COL pe chalta he  #NOTE Here the dimension and that of where to substitute
p = cv2.cvtColor(parrotface,cv2.COLOR_BGR2RGB) #just for more visibility

img0[155:355,0:400] = p

cv2.imshow("Merged Image",img0)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

`Blending of two or more images`
Output = src1*alpha + src2*beta + gamma

alpha = weight of src1

beta = weight of src2

gamma = scalar 

Can be done using **cv2.addWeighted()** takes 5 parameters, img1,alpha,img2,beta,gamma

```python
import cv2

img1 = cv2.imread('parrot.jpg')
img2 = cv2.imread('apple.png')

# Blending the images with weights
img1 = cv2.resize(img1,(300,500))
img2 = cv2.resize(img2,(300,500)) #NOTE First here is (col,row)
add = cv2.addWeighted(
    img1, 
    0.7,
    img2,
    0.3,
    0
)

cv2.imshow('Blended image',add)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# `Background Removal`
Useful when we need to isolate the object from the background abd remove the noise from the image.

In openCV we can perform Background removal using combination of thresholding, erosion, dilation and bitwise operations.

```python
import cv2

img = cv2.imread('apple.png')
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Remove the background using a Threshold
# Using the median filter to reduce the noise of image
gray_img = cv2.medianBlur(gray_img, 5)

# a binary threshold is applied to the gray image using a threshold
ret,thresh = cv2.threshold(gray_img, 220 , 255, cv2.THRESH_BINARY_INV)

# The binary image is eroded to remove small objects and fill in small gaps using erode
mask = cv2.erode(thresh, None, iterations=2)

# The original image is combined with binary mask using bitwise_and
background_removed_img = cv2.bitwise_and(img, img, mask=mask)

cv2.imshow('background_removed_img', background_removed_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

`Good Question` : draw a rectangle on the image and circle with modes if you want to change the shape then press m to change the mode 

```python
import cv2
import numpy as np

drawing = False #true if mouse pressed
mode = True #if true draw rectangles, press m to toggle
ix,iy = -1,-1
def mouse_event(event,x,y,flags,param):
    global ix,iy,drawing,mode
    if event == cv2.EVENT_LBUTTONDOWN:  #fist TAP THE left mouse button
        drawing = True
        ix,iy = x,y
    elif event == cv2.EVENT_MOUSEMOVE:  #while mouse is pressed and being moved
        if drawing == True:
            if mode == True:
                cv2.rectangle(img,(ix,iy),(x,y),(0,255,0),-1)
            else:
                cv2.circle(img,(x,y),10,(0,0,255),-1)
    elif event == cv2.EVENT_LBUTTONUP: #when the left mouse button is released draw till that
        drawing = False
        if mode == True:
            cv2.rectangle(img,(ix,iy),(x,y),(0,255,0),-1)
        else:
            cv2.circle(img,(x,y),5,(0,0,255),-1)
img = np.zeros((512,512,3),np.uint8)
cv2.namedWindow("Image")
cv2.setMouseCallback("Image",mouse_event)
while True:
    cv2.imshow("Image",img)
    k = cv2.waitKey(1) & 0xFF
    if k == ord('m'):
        mode = not mode
    elif k == ord('q'):
        break
cv2.destroyAllWindows()
```

# `Bitwise Operators`
Used in image processing to manipulate the pixel values. Several pixels operations such as `bitwise_and`, `bitwise_or`, `bitwise_xor` are available.

```python
import cv2
import numpy as np

img1 = cv2.imread('parrot.jpg')
img2 = cv2.imread('apple.png')

img1 = cv2.resize(img1,(300,300))
img2 = cv2.resize(img2,(300,300))

# Create a mask by thresholding the image
gray_img = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
ret, mask = cv2.threshold(gray_img, 127, 255, cv2.THRESH_BINARY)
mask = cv2.bitwise_not(mask)

# perform a bitwise AND operation on images
bitwise_and = cv2.bitwise_and(img1, img2, mask=mask)

# perfrom a bitwise OR operation on images
bitwise_or = cv2.bitwise_or(img1, img2, mask=mask)

# perform a biwwise XOR operation on images
bitwise_xor = cv2.bitwise_xor(img1, img2, mask=mask)

# Perform a bitwise NOT operation on images
bitwise_not = cv2.bitwise_not(img1)

cv2.imshow("IMAGE-1",img1)
cv2.imshow("IMAGE-2",img2)
cv2.imshow("MASK",mask)
cv2.imshow("BITWISE-AND",bitwise_and)
cv2.imshow("BITWISE-OR",bitwise_or)
cv2.imshow("BITWISE-XOR",bitwise_xor)
cv2.imshow("BITWISE-NOT",bitwise_not)
cv2.waitKey(0)  
cv2.destroyAllWindows()
```
