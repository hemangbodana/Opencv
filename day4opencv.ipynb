# `Image thresholding techniques`
1. **Simple Thresholding**
    - This is the simplest form of thresholding, where each pixel in the image is compared with a fixed threshold value. If the pixel value is greater than the threshold, it is set to a maximum value. Otherwise, it is set to zero
    - It is not very useful because it is very sensitive to the threshold value.
2. **Adaptive Thresholding**
    - The threshold value is not fixed, but instead is calculated for each pixel based on the mean or Gaussian-weighted mean of its surrounding pixels.
    - It is very useful because it is less sensitive to the threshold value.
3. **Otsu's Thresholding**
    - This is a technique for **automatically calculating the threshold value based on the image histogram**. It assumes that there are two classes of pixels in the image - foreground and background - and **finds the threshold that maximizes the variance between these two classes**.

4. **Binary thresholding**
    - This is similar to simple thresholding, but instead of setting the pixels above the threshold to a maximum value, they are set to a specific value (usually 1) to create a binary image. This can be done using the cv2.threshold() function with the cv2.THRESH_BINARY flag

```python
# thresholding
import cv2

img = cv2.imread("parrot.jpg",0)
cv2.imshow("Original", img)
# 1. Simple thresholding
threshold_val = 100
max_val = 255

ret,thresh = cv2.threshold(
    img, # input image
    threshold_val, # threshold value
    max_val, # max value
    cv2.THRESH_BINARY # threshold type
)
cv2.imshow("Simple thresholding", thresh)

# 2. Adaptive thresholding: best threshold for a range of values
adaptive_thresh = cv2.adaptiveThreshold(
    img, # input image
    max_val, # max value
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C, # threshold type/method
    cv2.THRESH_BINARY_INV, # threshold type
    11, # block size
    2 # C value
)
cv2.imshow("Adaptive thresholding", adaptive_thresh)

# 3. Otsu's thresholding
ret,otsu_thresh = cv2.threshold(
    img, # input image
    0, # threshold value
    max_val, # max value
    cv2.THRESH_BINARY + cv2.THRESH_OTSU # threshold type
)
cv2.imshow("Otsu's thresholding", otsu_thresh)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

# `Finding and Drawing Contours`

Countours are the outlines of objects in an image. They are useful for many tasks, such as object detection, tracking, and shape analysis.


```python
import cv2
import numpy as np

# Load the image 
img = cv2.resize(cv2.imread('desk.jpeg',0),(480,720))

# Apply thresholding to that image
ret,thresh = cv2.threshold(
    img,
    100,  ## threshold value
    255,  ## max value
    cv2.THRESH_BINARY  ## threshold type
)

# FIND CONTOURS
contours, hierarchy = cv2.findContours(
    thresh,
    cv2.RETR_TREE,  ## retrieve contours
    cv2.CHAIN_APPROX_SIMPLE  ## approximation method
)

# DRAW CONTOURS
cv2.drawContours(img, contours, -1, (255, 255, 255), 2)
cv2.imshow("Original Image",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# `Detect Simple Geometric Shapes`
Steps:
1. Load the image
2. Convert the image to grayscale
3. Apply a blur to remove noise
4. Apply a threshold to create a binary image
5. Find contours in the binary image
6. Draw the contours on the original image

```python
import cv2

img = np.zeros((480,480,3),np.uint8)
cv2.rectangle(
    img,
    (100,100),
    (300,300),
    (0,0,255),
    3
)
cv2.circle(
    img,
    (300,350),
    20,
    (255,0,0),
    3
)
# APPLY canny edge detection
canny = cv2.Canny(img,50,150,apertureSize=3)
cv2.imshow("Canny Edge Detection",canny)
# Find Contours in the edge image
contours,hirearchy = cv2.findContours(
    canny,
    cv2.RETR_TREE,
    cv2.CHAIN_APPROX_SIMPLE
)

# loop through the contour and check if it matches a circle,square, or traingle
for cnt in contours:
    # Calculate the perimeter of each cnt
    perimeter = cv2.arcLength(cnt,True)
    # approzimate the shape of the contour
    approx = cv2.approxPolyDP(cnt,0.04*perimeter,True)
    # Calculate number of vertices of the approzimated shape
    num_vertices = len(approx)
    
    if num_vertices == 3:
        cv2.drawContours(
            img,[approx],0,(0,255,255),2
        )
        cv2.putText(
            img,
            "Triangle",
            tuple(approx[0][0]),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.5,
            (255,255,255),
            2
        )
    elif num_vertices == 4:
        # Could be rectangle or square 
        x,y,w,h = cv2.boundingRect(cnt)
        aspect_ratio = float(w)/h
        if 0.95<=aspect_ratio<=1.05:
            cv2.drawContours(
                img,[approx],0,(0,0,255),2
            )
            cv2.putText(
                img,
                "Square",
                tuple(approx[0][0]),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,(0,0,255),2
            )
        else:
            cv2.drawContours(img,[approx],0,(0,255,255),2)
            cv2.putText(
                img,
                "Rectangle",
                tuple(approx[0][0]),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,(255,255,0),2
            )
    else:
        cv2.drawContours(img,[approx],0,(255,0,0),2)
        cv2.putText(
            img,
            "Circle",
            tuple(approx[0][0]),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.5,(255,0,0),2
        )



cv2.imshow("Original Image",img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

# `Template Matching`

Template matching is a technique used in computer vision to find a template (a
small image) within a larger image. OpenCV provides functions that can be used
to perform template matching in images.

The basic idea behind template matching is to slide a template image over a
larger image and compare the pixel values in the template image with the
corresponding pixels in the larger image. This comparison can be performed using two methods `Sum of Squared Differences (SSD)` correlation coefficient or `Normalized Cross-correlation (NCC)`.

-  cv2.matchTemplate()

    Parameters -> input_img and template_img
    Returns -> a response image that shows the similarity between the template and the input image.
    - Several mtds for matching the templates
      1. Sum of squared Difference `(cv2.TM_SQDIFF)`-> calculates the sum of squared differences between the template and the
input image at each pixel location. The **minimum value in the  resulting response image corresponds to the location of the best match.**
      2. Normalized Sum of Squarred differnces `(cv2.TM_SQDIFF_NORMED)`-> similar to the sum of squared differences method, but the resulting response image is **normalized to have values between 0 and 1**. The minimum value in the normalized response image corresponds to the location of the best match.

      3.  Cross-correlation `(cv2.TM_CCORR)`: This method calculates the **crosscorrelation between the template and the input image at each pixel location**. The **maximum value in the resulting response image corresponds** to the location of the best match.
      
      4. Normalized cross-correlation `(cv2.TM_CCORR_NORMED)`: This method is similar to the **cross-correlation method, but the resulting response image is normalized to have values between 0 and 1**. The **maximum value in the normalized response image corresponds to the location of the best match**.
      5. Correlation coefficient `(cv2.TM_CCOEFF)`: This method calculates the **correlation coefficient** between the template and the input image at each pixel location. The **maximum value in the resulting response image corresponds to the location of the best match**.
      
      6. Normalized correlation coefficient `(cv2.TM_CCOEFF_NORMED)`: This method is similar to the correlation coefficient method, but the resulting response image is **normalized to have values between -1 and 1**. The **maximum value in the normalized response image corresponds to the location of the best match**.

```python
# For image
import cv2
import numpy as np

small = cv2.resize(
    cv2.imread('small.jpg'),
    (300, 300)
)
img = cv2.imread('large.jpg')
tem_h, tem_w, _ = small.shape

result = cv2.matchTemplate(
    img,
    small,
    cv2.TM_CCORR_NORMED
)

# Find the BEST match
min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(result)

threshold_val = 0.8

if max_val >= threshold_val:

    top_left = max_loc

    bottom_right = (
        top_left[0] + tem_w,
        top_left[1] + tem_h
    )

    cv2.rectangle(
        img,
        top_left,
        bottom_right,
        (0, 0, 255),
        2
    )

    cv2.putText(
        img,
        f"Match: {max_val:.2f}",
        (top_left[0], top_left[1] - 10),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (0, 0, 255),
        2
    )

cv2.imshow('Live', img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
from typing import Tuple


def portrait_blur_pipeline(
    image: np.ndarray,
    face_box: Tuple[int, int, int, int],
    blur_ksize: int = 45
) -> np.ndarray:
    """
    Applies photographic portrait bokeh background blur around a detected face.
    face_box: (x, y, w, h)
    """
    x, y, w, h = face_box
    img_h, img_w = image.shape[:2]

    # 1. Apply heavy background blur across the entire frame
    ksize = blur_ksize if blur_ksize % 2 == 1 else blur_ksize + 1
    blurred_background = cv2.GaussianBlur(
        image, (ksize, ksize), 0
    )

    # 2. Construct feathered elliptical mask centered on face
    mask = np.zeros((img_h, img_w), dtype=np.float32)

    center = (
        int(x + w / 2),
        int(y + h / 2)
    )

    axes = (
        int(w * 0.75),
        int(h * 0.95)
    )

    cv2.ellipse(
        mask,
        center,
        axes,
        angle=0,
        startAngle=0,
        endAngle=360,
        color=1.0,
        thickness=-1
    )

    # Feather mask
    mask_feathered = cv2.GaussianBlur(
        mask,
        (51, 51),
        0
    )

    alpha = np.expand_dims(
        mask_feathered,
        axis=2
    )

    # 3. Alpha compositing
    composite = (
        image.astype(np.float32) * alpha
        +
        blurred_background.astype(np.float32) * (1.0 - alpha)
    )

    return np.clip(
        composite,
        0,
        255
    ).astype(np.uint8)


def run_live_portrait_blur():
    """
    Captures webcam video, detects a face, and applies
    portrait background blur in real time.
    """

    # Load OpenCV's Haar Cascade face detector
    face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades +
    "haarcascade_frontalface_default.xml"
)


    if face_cascade.empty():
        raise RuntimeError("Could not load Haar Cascade face detector.")

    # Open default webcam
    cap = cv2.VideoCapture(0)

    if not cap.isOpened():
        raise RuntimeError("Could not open webcam.")

    print("Live portrait blur started.")
    print("Press 'q' to quit.")

    while True:
        # --------------------------------------------------
        # 1. Capture frame
        # --------------------------------------------------
        ret, frame = cap.read()

        if not ret:
            print("Failed to read frame from webcam.")
            break

        # --------------------------------------------------
        # 2. Mirror the webcam
        # --------------------------------------------------
        frame = cv2.flip(frame, 1)

        # --------------------------------------------------
        # 3. Convert frame to grayscale for face detection
        # --------------------------------------------------
        gray = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2GRAY
        )

        # --------------------------------------------------
        # 4. Detect faces
        # --------------------------------------------------
        faces = face_cascade.detectMultiScale(
            gray,
            scaleFactor=1.1,
            minNeighbors=5,
            minSize=(80, 80)
        )

        # --------------------------------------------------
        # 5. Apply portrait blur
        # --------------------------------------------------
        output = frame

        if len(faces) > 0:

            # Use the largest detected face
            face_box = max(
                faces,
                key=lambda box: box[2] * box[3]
            )

            # Call your pipeline
            output = portrait_blur_pipeline(
                frame,
                tuple(face_box),
                blur_ksize=45
            )

            # Optional: draw face detection rectangle
            x, y, w, h = face_box

            cv2.rectangle(
                output,
                (x, y),
                (x + w, y + h),
                (0, 255, 0),
                2
            )

        # --------------------------------------------------
        # 6. Display result
        # --------------------------------------------------
        cv2.imshow(
            "Live Portrait Blur",
            output
        )

        # --------------------------------------------------
        # 7. Exit when Q is pressed
        # --------------------------------------------------
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break

    # ------------------------------------------------------
    # 8. Cleanup
    # ------------------------------------------------------
    cap.release()
    cv2.destroyAllWindows()


run_live_portrait_blur()
```
