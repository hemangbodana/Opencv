**Task:** Load an image and print its shape, dtype, and the pixel value at (50, 50).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Loading an image
img = cv2.imread('sampleimg.png')
print("image pixels",img)
print("Loaded image shape:(height,width,channel)", img.shape)
print("Loaded image dtype:", img.dtype)
print("The pixel value at (50,50) is:", img[50,50])
```

```python
cv2.imwrite('output.png', img)
```

```python
cv2.imshow("Output",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
cv2.imshow('image_rgb',img_rgb)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

cap = cv2.VideoCapture(0)
fourcc = cv2.VideoWriter_fourcc(*'XVID')
out = cv2.VideoWriter("out.avi",fourcc,20.0,(640,480))
while True:
    ret,frame = cap.read()
    if not ret:
        break
    cv2.imshow("Video",frame)
    out.write(frame)
    
    if cv2.waitKey(1)== ord('q'):
        break
    
cap.release()
out.release()
cv2.destroyAllWindows()
    
```

```python
import cv2 
import numpy as np

img = np.zeros((500,500,3),np.uint8)
cv2.imshow('img',img)
cv2.line(img,(0,0),(100,320),(255,0,0),3)
cv2.rectangle(img,(120,120),(200,200),(0,255,0),3)
cv2.circle(img,(220,220),50,(0,0,255),3)
pts = np.array([[300,200],[300,400],[500,200],[500,400]])
cv2.polylines(img,[pts],True,(255,255,255),3)
cv2.putText(img,'Hello',(0,500),cv2.FONT_HERSHEY_SIMPLEX,2,(0,255,255),3)

cv2.imshow('img',img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
